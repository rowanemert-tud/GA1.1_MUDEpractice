```python
def add(a, b):
    return a + b
```

```python
print(add(2, 3))
print(add(5, 7))
print(add(-2, 4))
```

```python
def sum_list(values):
    total = 0
    for value in values:
        total = add(total, value)
    return total
```

```python
print(sum_list([1, 2, 3]))
print(sum_list([5, 10, 15]))
print(sum_list([-2, 4, 6]))
```
